```python
# 1.4 4. Check Even or Odd
# Concept: Conditional statements.

num = int(input("Enter the number:"))
if num % 2 == 0:
    print("Even")
else:
    print("odd")
```

```python
# 1.6 6. Factorial of a Number
# Concept: Looping.

n = int(input("Enter the number:"))
fact = 1
for i in range(1, n+1):
    fact *= i
    print(fact)
```

```python
# 1.7 7.FibonacciSeries
# Concept: Iterationandvariables.
n=11
a, b = 0, 1
for i in range(n):
    print(a, end=" ")
    a, b = b, a+b
```

```python
# 1.9 9.PalindromeCheck
# Concept: Stringcomparison

p = "madam"
print("palindrome" if p == p[::-1] else "not")
```

```python
print("Prime numbers from 1 to 100:")

for n in range(2, 101):
    for i in range(2, n):
        if n % i == 0:
            break
    else:
        print(n, end=" ")
```
